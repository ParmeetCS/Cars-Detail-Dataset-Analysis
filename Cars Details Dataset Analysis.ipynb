# Cars Details Dataset Analysis

### Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Understanding Dataset

```python
df=pd.read_csv("CAR DETAILS.csv")
df
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

### Data Cleaning

```python
# To do list-
# 1.Dropping Null Values
# 2.Dropping Duplicate Values
# 3.Changing Data Type of km_driven to "int"
# 4.Droping Unnecessary Columns
```

```python
## Dropping Null Values
df.dropna()
```

```python
## Fill Null Values
df.fillna(df["km_driven"].mean(),inplace=True)
```

```python
## Dropping Duplicate Values
df.drop_duplicates(inplace=True)
```

```python
## Changing Data Type of km_driven
df["km_driven"]=df["km_driven"].astype(int)
```

```python
## Dropping Unnecessary Columns
df.drop(columns="seller_type",inplace=True)
df.drop(columns="transmission",inplace=True)
```

```python
df
```

```python
df.info()
```

```python
## Capitalize the first letter of Columns
new_columns=[]
for columns in df.columns:
    new_columns.append(columns.capitalize())
```

```python
df.columns=new_columns
```

```python
df
```

```python
#Extracting Minimum Price
df[df["Selling_price"]==df["Selling_price"].min()]
```

```python
#Extracting Maximum Price
df[df["Selling_price"]==df["Selling_price"].max()]
```

```python
#Description of Dataset
df.describe()
```

# Data Visualization of Dataset

### Visualizing number of cars under each fuel type

```python
sns.countplot(x="Fuel",data=df)
plt.ylabel("No.of cars")
plt.xlabel("Fuel Type")
plt.show()
```

### Selling Price vs Year

```python
selling_price_year=df.groupby("Year")["Selling_price"].mean().astype(int)
selling_price_year
```

```python
plt.plot(selling_price_year.index,selling_price_year.values)
plt.ylabel("Average Selling Price")
plt.xlabel("Year")
plt.show()
```
